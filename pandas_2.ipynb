## Cleaning data (Droping columns)

```python
practice_list = [
    {'Item': 'Apple', 'Qty': 10, 'Price': 0.5},
    {'Item': 'Banana', 'Qty': 5, 'Price': 0.3},
    {'Item': 'Cherry', 'Qty': 20, 'Price': 0.2},
    {'Item': 'Date', 'Qty': 7, 'Price': 0.4},
]
practice_df = pd.DataFrame(practice_list, index=['p1', 'p2', 'p3', 'p4'])
practice_df


#axis = 0 (drop row)
print(practice_df.drop("p1"))
print (practice_df)

print ()
#axis = 1 (drop column)
print(practice_df.drop("Qty", axis=1))
print (practice_df)
```

```python
practice_null_df = pd.DataFrame({
    'Customer': ['Alice', None, 'Charlie', 'David'],
    'Age': [25, pd.NA, 30, 22],
    'City': ['New York', 'Los Angeles', None, 'Chicago'],
    'SignupDate': [pd.Timestamp('2021-01-01'), pd.NaT, pd.Timestamp('2021-03-15'), None],
    'Active': [True, pd.NA, False, None],
    'Spend': [100.0, None, 50.5, 0.0]
}, index=['n1', 'n2', 'n3', 'n4'])

print(practice_null_df)
print("========")
print(practice_null_df.isnull().sum())

print()
print("========")

#remove rows with null values
print (practice_null_df.dropna())

print()
print("========")

# remove columns with null values 

print (practice_null_df.dropna(axis=1))

print()
print("========")

print (practice_null_df.dropna(subset=['Customer']))


print()
print("========")


```

## Filling in Null Values

```python

# print (practice_null_df.fillna(0))
# print()
# print("========")

print (practice_null_df['Customer'].fillna('Unknown'))
print()
print("========")

print (practice_null_df['Age'].fillna(practice_null_df['Age'].mean()))
print()
print("========")

print (practice_null_df['City'].fillna('Unknown'))
print()
print("========")

print (practice_null_df['SignupDate'].fillna('Unknown'))
print()
print("========")

print (practice_null_df['Active'].fillna('Unknown'))
print()
print("========")


print (practice_null_df['Spend'].fillna(0))
print()
print("========")

```
